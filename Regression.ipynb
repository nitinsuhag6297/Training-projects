```python
import numpy as np
from sklearn.linear_ model import LinearRegression
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([30,40,50,60,70])
model=LinearRegression()
model.fit(X,y)
prediction=model.predict([[6]])
print("Slope(Coefficient):",model.coef_)
print("Intercept:",model.intercept_)
print("Predicted Salary for 6 years of experience:",prediction[0])
```

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 5, 10, 17, 26])
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
model = LinearRegression()
model.fit(X_poly, y)
new_X = np.array([[6]])
new_X_poly = poly.transform(new_X)
prediction = model.predict(new_X_poly)
print("Predicted output for x = 6:", prediction[0])
```

```python
import numpy as np
import matplotlib.pyplot as plt

rolls = np.random.randint(1, 7, 1000)
outcomes = np.arange(1, 7)
experimental = []
for i in outcomes:
    probability = np.sum(rolls == i) / 1000
    experimental.append(probability)
theoretical = [1/6] * 6

for i in range(6):
    print("Outcome", outcomes[i],
          "Experimental:", experimental[i],
          "Theoretical:", theoretical[i])

x = np.arange(6)
width = 0.35

plt.bar(x - width/2, experimental, width, label="Experimental")
plt.bar(x + width/2, theoretical, width, label="Theoretical")

plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.title("Experimental vs Theoretical Probability")
plt.xticks(x, outcomes)
plt.legend()
plt.show()
```
