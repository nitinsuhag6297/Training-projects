**Answer:**

The education column contained 60 missing values, while the other columns had no missing values initially. I filled the missing education values using the most frequent category. I also identified five negative age values and five negative income values as invalid and replaced them with their respective median values. No duplicate rows were found. The median was used because it is less affected by extreme values.

```python

print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nNegative age values:", (df["age"] < 0).sum())
print("Negative income values:", (df["income"] < 0).sum())
```

```python

# 1. Replace invalid negative values with NaN
df.loc[df["age"] < 0, "age"] = np.nan
df.loc[df["income"] < 0, "income"] = np.nan

# 2. Fill missing numerical values with median
df["age"] = df["age"].fillna(df["age"].median())
df["income"] = df["income"].fillna(df["income"].median())

# 3. Fill missing education values with the most frequent category
df["education"] = df["education"].fillna(df["education"].mode()[0])

# 4. Verify the cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nNegative age values:", (df["age"] < 0).sum())
print("Negative income values:", (df["income"] < 0).sum())
```

```python

# Prepare the dataset for machine learning

# Remove customer_id because it is only an identifier
df = df.drop(columns=["customer_id"])

# Convert the target column into numerical values
# Yes = 1 means defaulted, No = 0 means repaid
df["loan_default"] = df["loan_default"].map({
    "Yes": 1,
    "No": 0
})

# Convert categorical columns into numerical dummy variables
# drop_first=True avoids redundant dummy columns
df = pd.get_dummies(df, drop_first=True, dtype=int)

# Separate input features (X) and target variable (y)
X = df.drop("loan_default", axis=1)
y = df["loan_default"]

# Display the dimensions and verify the prepared data
print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFirst five rows of features:")
display(X.head())

print("\nTarget values:")
print(y.value_counts())
```

**Answer:**

A Decision Tree in scikit-learn cannot directly use text columns because the algorithm requires numerical input features. Categorical text values must first be converted into numbers using encoding techniques such as one-hot encoding. This allows the model to process the data and make predictions.

```python

# Split the dataset into training and test sets

from sklearn.model_selection import train_test_split

# Split 80% data for training and 20% for testing
# stratify=y maintains the proportion of both target classes
# random_state=42 ensures the same split every time
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# Print the size of each dataset
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

**Answer:**

The dataset was split into 80% training data and 20% testing data. The training set contains 800 records, while the test set contains 201 records. We keep the test set separate to evaluate the model on unseen data and check how well it predicts loan defaults for new customers.

```python

# Train a Decision Tree Classifier

from sklearn.tree import DecisionTreeClassifier

# Create the Decision Tree model with default settings
dt = DecisionTreeClassifier(random_state=42)

# Train the model using the training data
dt.fit(X_train, y_train)

# Predict the target values for training and testing data
y_train_pred = dt.predict(X_train)
y_test_pred = dt.predict(X_test)

# Calculate accuracy on the training data
train_accuracy = dt.score(X_train, y_train)

# Calculate accuracy on the testing data
test_accuracy = dt.score(X_test, y_test)

# Display both accuracy scores
print("Training Accuracy:", train_accuracy)
print("Testing Accuracy:", test_accuracy)
```

The Decision Tree Classifier achieved 100% training accuracy and approximately 77.61% testing accuracy. The model performed perfectly on the training data but achieved lower accuracy on unseen test data. This difference suggests that the model may be overfitting. We can try limiting the tree's depth to improve its generalization.

```python

# Import metrics to evaluate the model's performance
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Display the confusion matrix
cm = confusion_matrix(y_test, y_test_pred)
print("Confusion Matrix:")
print(cm)

# Print the classification report
print("\nClassification Report:")
print(classification_report(y_test, y_test_pred, zero_division=0))

# Calculate individual evaluation metrics
print("Accuracy:", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_test_pred, zero_division=0))
print("F1-Score:", f1_score(y_test, y_test_pred, zero_division=0))
```

The Decision Tree Classifier achieved 77.61% accuracy on the test dataset. The confusion matrix shows that the model correctly identified 143 non-defaulters and 13 defaulters. However, it missed 24 actual defaulters. The precision for class 1 was 38.24%, while its recall was 35.14% and F1-score was 36.62%. This indicates that the model is better at predicting non-defaulters than identifying customers who may default. Therefore, accuracy alone is not sufficient to evaluate this model.

```python

# Import libraries for plotting the confusion matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Create a heatmap to visualize correct and incorrect predictions
plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Default", "Default"],
    yticklabels=["No Default", "Default"]
)

# Add a title and axis labels
plt.title("Decision Tree - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

# Display the plot
plt.show()
```

The confusion matrix heatmap provides a visual representation of the model's predictions. It shows that the model correctly predicted 143 non-defaulters and 13 defaulters. However, it incorrectly classified 21 non-defaulters as defaulters and missed 24 actual defaulters. The heatmap helps us understand the types of errors made by the model and shows that identifying defaulters is a major challenge.

```python

# Import the Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

# Create a Decision Tree with a maximum depth of 5
dt_limited = DecisionTreeClassifier(max_depth=5, random_state=42)

# Train the model using the training data
dt_limited.fit(X_train, y_train)

# Make predictions on training and testing data
y_train_pred_limited = dt_limited.predict(X_train)
y_test_pred_limited = dt_limited.predict(X_test)

# Calculate accuracy for both datasets
print("Training Accuracy:", dt_limited.score(X_train, y_train))
print("Testing Accuracy:", dt_limited.score(X_test, y_test))
```

After limiting the Decision Tree's maximum depth to 5, the training accuracy decreased from 100% to 86.63%, while the testing accuracy increased from 77.61% to 82.59%. This suggests that limiting the tree's depth helped reduce overfitting and improved its performance on unseen data. Therefore, the limited-depth Decision Tree performed better in terms of test accuracy.

```python

# Import evaluation metrics
from sklearn.metrics import classification_report, confusion_matrix

# Display the confusion matrix for the improved model
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_test_pred_limited))

# Display precision, recall, and F1-score
print("\nClassification Report:")
print(classification_report(y_test, y_test_pred_limited, zero_division=0))


```

The limited-depth Decision Tree achieved 82.59% testing accuracy. Its confusion matrix shows that it correctly classified 159 non-defaulters and 7 defaulters. The model's precision for the default class was approximately 58%, but its recall was only 18.92%. This means that it missed most of the actual defaulters. Although the overall accuracy improved, the model still needs improvement in identifying customers who may default. Therefore, precision, recall, and F1-score should also be considered when evaluating the model.

```python

# Import pandas to create a comparison table
import pandas as pd

# Store the accuracy scores of both models
comparison = pd.DataFrame({
    "Model": ["Original Decision Tree", "Limited-Depth Decision Tree"],
    "Training Accuracy": [
        dt.score(X_train, y_train),
        dt_limited.score(X_train, y_train)
    ],
    "Testing Accuracy": [
        dt.score(X_test, y_test),
        dt_limited.score(X_test, y_test)
    ]
})

# Display the comparison table
display(comparison)
```
