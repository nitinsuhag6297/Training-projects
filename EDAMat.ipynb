```python
 #IMPORTED REQUIRED LIBRARIES
 import numpy as np
 import pandas as pd
 import matplotlib.pyplot as plt
 import seaborn as sns
```

```python
#load the dataset
df=pd.read_csv("/content/salary_dataset (1) (1).csv")
df.head()
df.info()
df.describe()
```

```python
#checking for salary values
df.isnull().sum()
```

```python
#Univariate Analysis
#Distribution of employees by salary
plt.figure(figsize=(8,6))

sns.histplot(df['Salary'], bins=10, kde=True)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#Experience Distribution
plt.figure(figsize=(8,5))

sns.histplot(df['Experience'], bins=10, kde=True)

plt.title("Experience Distribution")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#Number of employees by jobrole
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="JobRole")

plt.title("Number of Employees by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
#Salary by JobRole
#Bi-variate analysis(including two variables)
plt.figure(figsize=(10,5))

sns.barplot(data=df, x="JobRole", y="Salary")

plt.title("Average Salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.show()
```

```python
#Salary by Education
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Education", y="Salary")

plt.title("Average Salary by Education")
plt.xlabel("Education")
plt.ylabel("Average Salary")
plt.show()
```

```python
#Salary by Location
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Location", y="Salary")

plt.title("Average Salary by Location")
plt.xlabel("Location")
plt.ylabel("Average Salary")
plt.show()
```

```python
#Experience vs Salary
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Education", y="Salary")

plt.title("Experience vs Salary")
plt.xlabel("Years of experience")
plt.ylabel("Salary")
plt.show()
```

```python
#With Regression Line
plt.figure(figsize=(8,5))

sns.regplot(data=df, x="Experience", y="Salary")

plt.title("Experience vs salary with Trend Lines")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```
